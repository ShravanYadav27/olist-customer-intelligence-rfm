```python
analysis_date = customer_data["order_purchase_timestamp"].max() + pd.Timedelta(days=1)

analysis_date
```

```python
frequency = customer_data.groupby("customer_unique_id")["order_id"].nunique()

frequency.head()
```

```python
monetary = customer_data.groupby("customer_unique_id")["payment_value"].sum()

monetary.head()
```

```python
rfm = pd.DataFrame({
    "Recency": recency,
    "Frequency": frequency,
    "Monetary": monetary
})

rfm.head()
```

```python
repeat_customers = (rfm["Frequency"] > 1).sum()

repeat_rate = (repeat_customers / len(rfm)) * 100

print("Repeat Customers:", repeat_customers)
print("Repeat Customer Rate:", round(repeat_rate, 2), "%")
```

```python
def frequency_score(x):
    if x == 1:
        return 1
    elif x == 2:
        return 2
    elif x <= 4:
        return 3
    else:
        return 4

rfm["F_Score"] = rfm["Frequency"].apply(frequency_score)
```

```python
rfm["F_Score"].value_counts().sort_index()
```

```python
def recency_score(x):
    if x <= 114:
        return 4
    elif x <= 219:
        return 3
    elif x <= 346:
        return 2
    else:
        return 1

rfm["R_Score"] = rfm["Recency"].apply(recency_score)
```

```python
rfm["R_Score"].value_counts().sort_index()
```

```python
def monetary_score(x):
    if x <= 63.06:
        return 1
    elif x <= 107.78:
        return 2
    elif x <= 182.56:
        return 3
    else:
        return 4

rfm["M_Score"] = rfm["Monetary"].apply(monetary_score)
```

```python
rfm["M_Score"].value_counts().sort_index()
```

```python
rfm.head(10)
```

```python
rfm = rfm[[
    "Recency",
    "Frequency",
    "Monetary",
    "R_Score",
    "F_Score",
    "M_Score"
]]

rfm.head()
```

```python
rfm["RFM_Code"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

rfm.head()
```

```python
pd.crosstab(
    rfm["R_Score"],
    rfm["F_Score"]
)
```

```python
def assign_segment(row):
    R = row["R_Score"]
    F = row["F_Score"]
    M = row["M_Score"]

    # Repeat customers
    if R >= 3 and F >= 3 and M >= 3:
        return "Champions"

    elif R >= 3 and F >= 2:
        return "Loyal Customers"

    elif R <= 2 and F >= 2:
        return "At Risk"

    # One-time customers
    elif R == 4 and F == 1:
        return "Recent Customers"

    elif F == 1 and M == 4:
        return "High-Value One-Time"

    elif R == 3 and F == 1:
        return "Promising"

    elif R == 2 and F == 1:
        return "Needs Attention"

    else:
        return "Inactive"


rfm["Segment"] = rfm.apply(assign_segment, axis=1)
```

```python
rfm["Segment"].value_counts()
```

```python
segment_summary = rfm.groupby("Segment").agg(
    Customers=("Frequency", "size"),
    Avg_Recency=("Recency", "mean"),
    Avg_Frequency=("Frequency", "mean"),
    Avg_Monetary=("Monetary", "mean"),
    Total_Revenue=("Monetary", "sum")
).reset_index()

segment_summary
```

```python
segment_summary["Customer_%"] = (
    segment_summary["Customers"] / segment_summary["Customers"].sum() * 100
)

segment_summary["Revenue_%"] = (
    segment_summary["Total_Revenue"] / segment_summary["Total_Revenue"].sum() * 100
)

segment_summary[[
    "Segment",
    "Customers",
    "Customer_%",
    "Avg_Recency",
    "Avg_Frequency",
    "Avg_Monetary",
    "Total_Revenue",
    "Revenue_%"
]].sort_values("Total_Revenue", ascending=False)
```

```python
segment_actions = {
    "Champions": "Retain with VIP rewards and exclusive offers",
    "Loyal Customers": "Strengthen loyalty and encourage higher spend",
    "Recent Customers": "Encourage a second purchase",
    "Promising": "Nurture with personalized recommendations",
    "High-Value One-Time": "Target for repeat purchase and loyalty conversion",
    "Needs Attention": "Re-engage before they become inactive",
    "At Risk": "Run targeted win-back campaigns",
    "Inactive": "Use selective reactivation campaigns"
}

rfm["Recommended_Action"] = rfm["Segment"].map(segment_actions)
```
